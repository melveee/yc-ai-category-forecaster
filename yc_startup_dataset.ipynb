# Assign categories to AI candidates

```python
def assign_categories(row):
    tags = {
        tag.strip().casefold()
        for tag in str(row["Tags"]).split(";")
    }

    matched_categories = []

    for category, rule_tags in category_rules.items():
        if any(tag.casefold() in tags for tag in rule_tags):
            matched_categories.append(category)

    return matched_categories


ai_candidates_v2 = ai_candidates_v2.copy()

ai_candidates_v2["AI Categories"] = ai_candidates_v2.apply(
    assign_categories,
    axis=1
)

print(
    ai_candidates_v2[
        ["Company Name", "AI Categories"]
    ].head(15).to_string(index=False)
)
```

```python

# Use category shares, excluding incomplete 2026
share_data = yearly_category_share.loc[
    yearly_category_share.index < 2026,
    yearly_category_counts.columns
].copy()

# Evaluate a 3-year moving-average baseline
results = []

for test_year in [2024, 2025]:
    train_data = share_data.loc[
        share_data.index < test_year
    ]

    actual = share_data.loc[test_year]
    predicted = train_data.tail(3).mean()

    for category in share_data.columns:
        results.append({
            "Test Year": int(test_year),
            "Category": category,
            "Actual Share (%)": actual[category],
            "Predicted Share (%)": predicted[category],
            "Absolute Error (pp)": abs(
                actual[category] - predicted[category]
            )
        })

backtest_results = pd.DataFrame(results)

print(
    backtest_results.round(2).to_string(index=False)
)

print("\nMean absolute error by test year:")
print(
    backtest_results.groupby("Test Year")[
        "Absolute Error (pp)"
    ].mean().round(2)
)
```

```python
naive_results = []

for test_year in range(2019, 2026):
    train_data = share_data.loc[
        share_data.index < test_year
    ]

    actual = share_data.loc[test_year]
    predicted = train_data.iloc[-1]

    for category in share_data.columns:
        naive_results.append({
            "Test Year": test_year,
            "Category": category,
            "Absolute Error (pp)": abs(
                actual[category] - predicted[category]
            )
        })

naive_backtest = pd.DataFrame(naive_results)

print("Naive MAE by year:")
print(
    naive_backtest.groupby("Test Year")["Absolute Error (pp)"]
    .mean()
    .round(2)
)

print("\nOverall Naive MAE:")
print(round(naive_backtest["Absolute Error (pp)"].mean(), 2))
```

```python
from sklearn.linear_model import LinearRegression
import numpy as np
import pandas as pd

trend_results = []

for test_year in range(2019, 2026):
    train_data = share_data.loc[
        share_data.index < test_year
    ]

    actual = share_data.loc[test_year]

    X_train = train_data.index.to_numpy().reshape(-1, 1)
    X_test = np.array([[test_year]])

    for category in share_data.columns:
        y_train = train_data[category].to_numpy()

        model = LinearRegression()
        model.fit(X_train, y_train)

        predicted = model.predict(X_test)[0]

        trend_results.append({
            "Test Year": test_year,
            "Category": category,
            "Absolute Error (pp)": abs(
                actual[category] - predicted
            )
        })

trend_backtest = pd.DataFrame(trend_results)

print("Linear-trend MAE by year:")
print(
    trend_backtest.groupby("Test Year")["Absolute Error (pp)"]
    .mean()
    .round(2)
)

print("\nOverall linear-trend MAE:")
print(round(trend_backtest["Absolute Error (pp)"].mean(), 2))
```

# identify the most defensible categories

```python
# Classify categories using historical momentum
# and forecast disagreement

def classify_category(row):
    momentum = row["Historical Change (pp)"]
    spread = row["Forecast Spread (pp)"]

    if momentum > 0 and spread <= 2:
        return "Positive momentum, lower disagreement"

    elif momentum > 0 and spread > 2:
        return "Positive momentum, higher disagreement"

    elif momentum <= 0 and spread <= 2:
        return "Declining/stable momentum, lower disagreement"

    else:
        return "Declining/stable momentum, higher disagreement"


forecast_summary["Category Assessment"] = (
    forecast_summary.apply(classify_category, axis=1)
)

display(
    forecast_summary[
        [
            "Average Forecast (%)",
            "Historical Change (pp)",
            "Forecast Spread (pp)",
            "Category Assessment",
        ]
    ].round(2)
)
```
